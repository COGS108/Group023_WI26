## Abstract

This study investigates the relative influence of academic effort and socioeconomic factors on student exam performance. Using exploratory data analysis, correlation analysis, and linear regression models, we examined how study time and parental education levels relate to exam scores. The results indicate that study time has a weak positive association with performance, while maternal education shows a stronger and statistically significant relationship. However, the overall explanatory power of the models remains low, suggesting that these variables alone do not fully account for variation in student outcomes. These findings highlight that academic performance is shaped by multiple interacting factors, with socioeconomic background playing a somewhat more prominent role than individual effort.

## Research Question

__Do socioeconomic factors outweigh an individuals academic effort in predicting a student's final exam score?__ <br>

We aim to determine whether students from low income backgrounds can compensate for their lower socioeconomic support with high study hours. In general, can a student from a poor family compete academically with a student from a wealthy family through hard work? We will seek to answer this question by performing linear regression and correlation analysis to compare effort-based variables (hours studied, attendance) to family advantages (family income, parental education level, access to resources).


## Background and Prior Work

There has been an ongoing debate on what factors make a student successful. The American dream tells us that if you work hard enough, you can achieve top grades no matter the background. But according to educational researchers, a student's family background is often a stronger predictor of success than their individual effort.<a name="cite_ref-1"></a><sup>1</sup> Using real world data, we wish to verify if hard work, like studying, can truly compete with advantages like family income. Through data science, we hope to check whether this traditional notion of success through hard work is factual or an unattainable myth.

Being raised in a wealthy family where their parents are highly educated more often than not provides an unequal advantage. Studies show that wealth translates directly into better learning environments and more abundant learning resources.<a name="cite_ref-2"></a><sup>2</sup> This includes minor details such as quiet comfortable places to study, faster and stabler internet, and parents or hired tutors who are readily available to help with their study. As such, wealthier students might study for similar hours compared to a low income student, but yield better grades for the same amount of work and time put in.

Although these advantages appear to be highly relevant to academic success, hard work is often still required in order to succeed. Analysis of real word data on study hours and attendance prove to still be statistically significant.<a name="cite_ref-3"></a><sup>3</sup> This suggests that an individual's choice also plays a large role in success. However, whether hard work can actually close the gap between wealth and family support is the question we are actually curious about. Can a student level their playing field simply by putting in more effort.

We must also examine the role of parents in their children's success. Research consistently proves that the parent's education level is the single largest factor in determining the success of their children.<a name="cite_ref-4"></a><sup>4</sup> Parents who hold degrees know the school system well enough to guide their children through the hurdles of college. This assistance is a significant advantage unrelated to their children's intelligence or work ethic. From comparing background factors like such against the student's own motivation, we can extrapolate the greater factor in attaining a higher final exam score.

### Dataset #1: Student Performance Factors (Kaggle)

**Description & Relevance**
This dataset contains survey results from over 6,000 students with variables on individual effort and background.
- **Exam_Score**: The target variable, representing the student's final grade (0-100).
- **Hours_Studied**: Number of hours studied per week
- **Attendance**: A percentage or count representing class attendance (higher is better).
- **Family_Income**: (Low, Medium, High) representing the socioeconomic status of the student's family.
- **Parental_Education_Level**: (High School, College) indicating the highest education level of the parents.
- **Access_to_Resources**: (Low, Medium, High) indicating access to educational materials and internet.

**Potential Issues & Bias**
The data comes from a survey by the students themselves, therefore there is the possibility that variables such as income or study hours may have response bias. Also, the data does not describe where or what type of schools (private or public) the survey was performed and the quality of the surveyed schools, so there may be additional biases. 

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm

# Loading first set of data
df_1 = pd.read_csv('data/00-raw/kaggle_data.csv')

print(f"Row x Columns: {df_1.shape}")

# Checking for Missing Data
print(f"Missing Values per Column: {df_1.isnull().sum()}")
df_1.head(5)
```

**Data contains 6607 samples, with missing data in Teacher_Quality, Physical_Activity, Parental Education**

#### Missing Data Analysis
- Parental_Education_Level is the only relevant variable with missing data to our research question, therefore we will check for correlation between missing data and Exam_Score to assure there is not systematic relationship with missingness.
- Other rows with missing values will be dropped, as the data is large enough compared to missing values (<5%) and the variables are not relevant to our research question.

```python
# Checking for systematic missingness 
df_1['missing_parent_ed'] = df_1['Parental_Education_Level'].isnull().astype(int)

# Find correlation between Parental_Education_Level and Exam_Score
corr_val = df_1['missing_parent_ed'].corr(df_1['Exam_Score'])
print(f"Correlation of Missing Parental_Education_Level vs Exam_Score: {corr_val:.2f}")

if abs(corr_val) < 0.05: 
    print("Result: Correlation below threshold of 0.05.")
    print("Conclusion: Negligible correlation, so the data is safe to drop.")
else:
    print("Result: Correlation above threshold of 0.05.")
    print("Conclusion: Significant correlation, so the data should be preserved.")

df_1 = df_1.drop(columns=['missing_parent_ed'])
```

**Test Results:** We checked with point-biserial correlation coefficient to determine that the data points regarding Parental_Education_Level were not missing completely at random, so the data is irrelevant to the student's ability, and is safe to drop.

```python
# Remove rows with missing values
df_1_clean = df_1.dropna().copy()

# Check for duplicate entries
duplicates = df_1_clean.duplicated().sum()

assert duplicates == 0, "Contains Duplicates"

# Plot boxplot to visualize data in case of impossible values
plt.figure(figsize = (8, 4))
sns.boxplot(x=df_1_clean['Exam_Score'])
plt.title('Exam Score Boxplot')
plt.show()
```

**Results**: Boxplot appears to only contain possible scores (0-100). The extra data points above 100 appears to suggest there is extra credit. We will proceed with this data.

```python
# Removing irrelevant variables and mapping strings of categorical vars to ints
cols_to_drop = ['Distance_from_Home', 'Physical_Activity', 'Peer_Influence', 
                'Sleep_Hours','Parental_Involvement', 'Learning_Disabilities', 
                'Previous_Scores', 'Gender', 'Extracurricular_Activities']

education_map = {
    'High School': 1,
    'College': 2,
    'Postgraduate': 3
}

intensity_map = {
    'Low': 1,
    'Medium': 2,
    'High': 3
}

binary_map = {
    'Yes': 1,
    'No': 0
}

school_map = {
    'Public': 0,
    'Private': 1
}

df_1_clean['Parental_Education_Level'] = df_1_clean['Parental_Education_Level'].map(education_map)

intensity_cols = ['Access_to_Resources', 'Motivation_Level', 'Family_Income', 'Teacher_Quality']
for col in intensity_cols:
    if col in df_1_clean.columns:
        df_1_clean[col] = df_1_clean[col].map(intensity_map)

df_1_clean['Internet_Access'] = df_1_clean['Internet_Access'].map(binary_map)
df_1_clean['School_Type'] = df_1_clean['School_Type'].map(school_map)

df_1_clean = df_1_clean.drop(columns=cols_to_drop)

# Saving Data to data/02-processed
df_1_clean.to_csv('data/02-processed/kaggle_dataset_cleaned.csv', index=False)
```

#### Correlation Between Variables

To better understand how different variables relate to one another, we compute a **correlation matrix**. Correlation measures the strength and direction of the relationship between two variables.

Values range from **-1 to 1**:
- **1** = strong positive relationship  
- **0** = no relationship  
- **-1** = strong negative relationship  

A heatmap visualization helps identify which variables are most strongly associated with **Exam_Score**. This allows us to quickly see whether effort-related variables or socioeconomic variables appear to have stronger relationships with student performance.

```python
# select only numeric variables
numeric_df = df_2_relevant.select_dtypes(include=['number'])

# compute correlation matrix
corr_matrix = numeric_df.corr()

plt.figure(figsize=(10,8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt=".2f"
)

plt.title("Correlation Matrix of Student Performance Variables")

plt.show()
```

The heatmap displays the correlation coefficients between all numeric variables in the dataset, allowing us to assess both the strength and direction of relationships. Darker colors indicate stronger correlations, while lighter colors indicate weaker or negligible relationships.

Similarly, socioeconomic indicators such as parental education (`Medu` ≈ 0.22, `Fedu` ≈ 0.15) show slightly stronger—but still modest—positive relationships with exam scores.

One notable relationship is the strong negative correlation between `Exam_Score` and `is_zero_score` (≈ -0.74), which is expected since this variable directly reflects whether a student received a zero score.

Overall, the heatmap suggests that no single variable strongly predicts exam performance. Both academic effort and socioeconomic factors show only weak to moderate associations, indicating that student outcomes are likely influenced by a combination of multiple interacting factors rather than a single dominant predictor.

## Analysis

In this section, we move beyond exploratory analysis to quantitatively evaluate how different factors influence exam performance. Specifically, we compare the predictive strength of academic effort (study time) and socioeconomic factors (such as parental education) using a regression model.

By fitting a linear regression model, we can estimate how much each variable contributes to changes in exam score while controlling for other variables. This allows us to determine whether academic effort or socioeconomic background has a stronger influence on student performance.

```python
y = df_2_relevant['Exam_Score']
X1 = sm.add_constant(df_2_relevant[['studytime']])
model1 = sm.OLS(y, X1).fit()
print(model1.summary())
```

```python
X2 = sm.add_constant(df_2_relevant[['Medu', 'Fedu']])
model2 = sm.OLS(y, X2).fit()
print(model2.summary())
```

**Results & Interpretation**

To evaluate the relative influence of academic effort and socioeconomic factors on exam performance, we fit two linear regression models.

The first model examined academic effort using study time as the predictor. The results show that study time has a small positive coefficient (β ≈ 0.53), indicating that increased study time is associated with slightly higher exam scores. However, this relationship is not statistically significant at the 0.05 level (p ≈ 0.052), and the model explains very little of the variance in exam scores (R² ≈ 0.010). This suggests that study time alone is a weak predictor of student performance.

The second model examined socioeconomic factors using parental education levels. Maternal education (Medu) has a positive and statistically significant relationship with exam score (β ≈ 0.84, p ≈ 0.002), indicating that higher maternal education is associated with higher student performance. In contrast, paternal education (Fedu) shows a small and non-significant effect (β ≈ 0.12, p ≈ 0.658), suggesting it does not meaningfully predict exam outcomes in this dataset.

The socioeconomic model explains more variation in exam scores than the effort model (R² ≈ 0.048 vs. 0.010), though the overall explanatory power remains modest. This indicates that while socioeconomic factors—particularly maternal education—have a stronger influence than study time, neither set of variables alone fully explains differences in student performance.

Overall, these results suggest that socioeconomic factors have a somewhat stronger predictive relationship with exam performance than academic effort, but both effects are relatively limited. This supports the conclusion that student outcomes are influenced by multiple interacting factors rather than a single dominant predictor.

## Discussion and Conclusion

Overall, this analysis explored how academic effort and socioeconomic factors relate to student exam performance. The results show that study time has a small positive relationship with exam scores, but this effect is weak and not statistically significant. In contrast, maternal education shows a stronger and statistically significant positive relationship with exam performance, while paternal education does not appear to have a meaningful effect. This suggests that certain aspects of socioeconomic background may play a more influential role than individual effort alone.

These findings connect back to the broader idea discussed in the background section that student performance is shaped by both individual behaviors and external factors. While effort, such as studying more, is important, it may not fully overcome differences in access to resources, support systems, or educational environments that are often tied to family background. The results extend this idea by showing that not all socioeconomic variables have the same impact, with maternal education standing out more clearly than paternal education.

When considering the results together, a key takeaway is that no single factor strongly predicts exam performance. Both the correlation analysis and regression models show relatively low explanatory power, meaning that student outcomes are likely influenced by a combination of many variables rather than one dominant predictor. This highlights the complexity of academic success and suggests that factors such as motivation, teaching quality, mental health, and access to resources may also play important roles but were not fully captured in this dataset.

There are also some limitations to this analysis. The dataset may not be fully representative, and the relationships observed are based on cross-sectional data, which means we cannot determine causation. Additionally, some important variables may be missing or imperfectly measured, which could weaken the observed relationships. The relatively low R² values in both models further indicate that much of the variation in exam scores remains unexplained.

If this investigation were to continue, future work could include incorporating additional variables such as socioeconomic status, school quality, or psychological factors. Using longitudinal data could also help better understand how these relationships change over time. Overall, this analysis suggests that while both effort and socioeconomic context matter, student performance is influenced by a complex interaction of multiple factors rather than a single cause.
